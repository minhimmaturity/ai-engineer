## 3. Tải mô hình pretrained
Dùng mô hình BERT-Tiny đã học phân loại SMS. Lần đầu cần Internet để tải trọng số (~18 MB).

```python

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    pipeline
)

model_id = "niru-nny/SMS_Spam_Detection"

tokenizer = AutoTokenizer.from_pretrained(
    model_id, force_download=True
)
model = AutoModelForSequenceClassification.from_pretrained(
    model_id, force_download=True
)

classifier = pipeline(
    "text-classification",
    model=model,
    tokenizer=tokenizer,
    device=-1
)

tests = [
    "Hey, are we still meeting for lunch?",
    "Please send me the lecture notes.",
    "URGENT! You won a FREE iPhone. Claim your prize now!",
    "Congratulations! Call this number to claim your cash reward.",
    "Your appointment is confirmed for tomorrow."
]

for msg in tests:
    print(msg)
    print(classifier(msg, top_k=None))
    print("-" * 50)

print("Đã khởi tạo mô hình!" if classifier is not None else "Chưa hoàn thành TODO")
```

## 4. Dự đoán một tin nhắn
Điền tin nhắn và gọi mô hình. Nhãn dự đoán được hiển thị trong `label`.

```python
message = "Congratulations! This is a free call to annouce that you won a free prize. Please contact us for get the instruction to receive the prize"

# TODO [ ] Dùng classifier dự đoán message
# Gợi ý: result = classifier(message)
result = classifier(message)

print(result)
```

## 5. Thử nhiều tin nhắn
Dự đoán cả danh sách. Sau đó **thay một câu** bằng tin nhắn do em tự nghĩ ra.

```python
messages = [
    "Good afternoon, I call you to say that we gonna have a meeting in 7p.m tonight, don't forget it",
    "Haha, I got nuts. Thanks dealer. That poker man",
    "Have you finished your homework yet"
]

"""
Chưa hiểu phần LABEL và accurancy score ở đây lắm, khi nào là score là 1.0,
phần nào ảnh hưởng đến accurrancy score?
Ví dụ như trong 1 message thì model dựa vào đâu để dự đoán, biết đó là message dựa trên nhưng thứ đã được train.
"""
# TODO [ ] Dự đoán toàn bộ danh sách
# Gợi ý: results = classifier(messages)
results = classifier(messages)

for message, prediction in zip(messages, results):
    print(message)
    print(prediction["label"], round(prediction["score"], 3))
    print("---")
```
