## 3. Tạo bảng dữ liệu

Tạo bảng từ điểm của 4 học viên.

```python
data = {
    "Tên": ["An", "Bình", "Chi", "Dũng"],
    "Điểm": [8, 4, 9, 6]
}

df = pd.DataFrame(data)
print(df)
# TODO [ ] Tạo bảng df bằng pd.DataFrame(data)
# TODO [ ] Hiển thị bảng bằng print(df)
```

## 4. Tính điểm trung bình và cao nhất

Điền tên hàm phù hợp vào dấu `...`. **Kết quả cần có:** 6.75 và 9.

```python
# TODO [ ] Điểm trung bình: dùng .mean()
trung_binh = df["Điểm"].mean()

# TODO [ ] Điểm cao nhất: dùng .max()
cao_nhat = df["Điểm"].max()

print("Trung bình:", trung_binh)
print("Cao nhất:", cao_nhat)
```

## 5. Lọc học viên đạt

**Yêu cầu:** Chỉ hiển thị những người có điểm từ 5 trở lên. **Kết quả:** An, Chi, Dũng.

```python
# TODO [ ] Điền điểm tối thiểu để đạt
dat = df[df["Điểm"] >= 5]
print(dat)
```

## 6. Tự kiểm tra

Chạy ô này sau khi hoàn thành các bài trên.

```python
# Không cần sửa ô này
kiem_tra = [
    ("Đã tạo bảng", "df" in globals() and isinstance(globals().get("df"), pd.DataFrame) and len(df) == 4),
    ("Điểm trung bình", "trung_binh" in globals() and abs(float(globals().get("trung_binh", 0)) - 6.75) < 1e-8),
    ("Điểm cao nhất", "cao_nhat" in globals() and globals().get("cao_nhat") == 9),
    ("Lọc học viên đạt", "dat" in globals() and list(globals().get("dat", pd.DataFrame()).get("Tên", [])) == ["An", "Chi", "Dũng"]),
]
for ten, dung in kiem_tra:
    print(("✅" if dung else "❌"), ten)
print(f"Hoàn thành: {sum(ok for _, ok in kiem_tra)}/{len(kiem_tra)}")
```

## 7. Bài thêm (không bắt buộc)

Tìm điểm thấp nhất bằng `.min()` và sắp xếp theo điểm giảm dần bằng `.sort_values()`.

```python
# TODO [ ] Tìm điểm thấp nhất
# TODO [ ] Sắp xếp bảng theo Điểm giảm dần
# Gợi ý: df.sort_values(by="Điểm", ascending=False)

lowest_score = df["Điểm"].min()

score_sort = df.sort_values(by="Điểm", ascending=False)
score_sort["STT"] = range(1, len(score_sort) + 1)
cols = ["STT"] + [col for col in score_sort.columns if col != "STT"]
score_sort = score_sort[cols]

print("Điểm thấp nhất" + ": " + str(lowest_score))
print(" ")
print("-----------------------------")
print("Sắp xếp điểm giảm dần")
print(" ")
print(score_sort.to_string(index=False))
```
